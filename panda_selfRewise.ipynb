```python
import pandas as pd
x = [3,4,5,6,7]
var = pd.Series(x,index=[1,2,3,4,5], dtype="float", name = "PYTHON")
print(var)
```

```python
print(type(var))
print(var[2])
```

```python
dic = {"name": ['python', 'c', 'c++', 'java'], "por":[11,45,33,75], "rank": [1,4,2,3]}
print("before converting to the Series form: ",dic)
var1 = pd.Series(dic)
print("After converting to the Series: ",var1)
```

```python
s1 =pd.Series(8, index=[1,2,3,4,5,6,7,8])
s2 = pd.Series(4, index=[1,2,3,4])
print(s1+s2)
```

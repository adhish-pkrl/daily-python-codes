```python
import matplotlib.pyplot as plt

# Monthly sales
total_sales_car = df.groupby('month')['sellingprice'].sum()

# Create figure
plt.figure(figsize=(12, 6))

# Plot line
plt.plot(
    total_sales_car.index,
    total_sales_car.values,
    marker='o',
    linestyle='-',
    linewidth=3,
    markersize=8,
    color='royalblue'
)

# Add grid
plt.grid(True, linestyle='--', alpha=0.5)

# Labels
plt.xlabel("Month", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)

# Title
plt.title(
    "Monthly Car Sales Analysis",
    fontsize=16,
    fontweight='bold',
    color='darkred'
)

# Display value on each point
for x, y in zip(total_sales_car.index, total_sales_car.values):
    plt.text(
        x,
        y,
        f'{y:,.0f}',
        ha='center',
        va='bottom',
        fontsize=8
    )

# Month ticks
plt.xticks(range(1, 13))

plt.tight_layout()
plt.show()
```

```python
total_sales_car = df.groupby('day_name')['sellingprice'].sum()
print(total_sales_car)
total_sales_car.plot(kind = "line", figsize=(10,6), marker = "o", title='weekly sales')
plt.xlabel("day_name")
plt.ylabel("Total sales")
plt.xticks(rotation= 45)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Correct day order
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

# Group sales by day
total_sales_car = df.groupby('day_name')['sellingprice'].sum()

# Reorder days
total_sales_car = total_sales_car.reindex(day_order)

# Create figure
plt.figure(figsize=(12, 6))

# Plot
plt.plot(
    total_sales_car.index,
    total_sales_car.values,
    marker='o',
    linewidth=3,
    markersize=8,
    color='royalblue'
)

# Add values above points
for x, y in zip(total_sales_car.index, total_sales_car.values):
    plt.text(x, y, f'{y:,.0f}', ha='center')

# Styling
plt.title(
    "Weekly Car Sales Analysis",
    fontsize=16,
    fontweight='bold'
)

plt.xlabel("Day of Week", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

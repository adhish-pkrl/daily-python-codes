```python
import matplotlib.pyplot as plt
import numpy as np

# Create figure
fig, ax = plt.subplots(figsize=(12, 6))

# Generate gradient colors
colors = plt.cm.viridis(np.linspace(0.2, 0.9, len(total_sales)))

# Create bar graph
bars = ax.bar(
    total_sales.index,
    total_sales.values,
    color=colors,
    edgecolor="white",
    linewidth=1
)

# Add values above each bar
ax.bar_label(
    bars,
    labels=[f"{value:,.0f}" for value in total_sales.values],
    padding=4,
    fontsize=9,
    fontweight="bold"
)

# Labels and title
ax.set_xlabel("Items", fontsize=12, fontweight="bold")
ax.set_ylabel("Total Sales", fontsize=12, fontweight="bold")
ax.set_title("Item-wise Sales Analysis", fontsize=16, fontweight="bold", pad=20)

# Rotate X-axis labels
plt.xticks(rotation=45, ha="right")

# Customize grid
ax.grid(axis="y", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Add spacing
ax.margins(y=0.15)

# Adjust layout
plt.tight_layout()
plt.show()
```

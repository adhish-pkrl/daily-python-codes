```python
import matplotlib.pyplot as plt
total_sales = df.groupby("Items")["Total sales"].sum()

print(total_sales)
```

```python
plt.figure(figsize=(10,6))
plt.bar(total_sales.index, total_sales.values)
plt.xlabel("Items")
plt.ylabel("Total sales")
plt.xticks(rotation = 35)
plt.show()
```
